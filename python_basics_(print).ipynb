4. mobile recharge calculator

```python
recharge_amount=float(input("Enter recharge_amount:"))
gst_percentage=float(input("Enter gst_percentage:"))
gst=recharge_amount*gst_percentage/100
final_amount=recharge_amount+gst
print("total:",final_amount)
```

5. temparature converter

```python
celsius=float(input("Enter temparature into celcius:"))
fahrenheit=(celsius*9/5)+32
print("temparature in farenheit:",fahrenheit)
```

6. simple intrest calculator

```python
principal = float(input("Enter principal amount: "))
rate = float(input("Enter rate of interest: "))
time = float(input("Enter time in years: "))

simple_interest = (principal * rate * time) / 100
total_amount = principal + simple_interest

print("Simple Interest = ₹", simple_interest)
print("Total Amount = ₹", total_amount)
```

7. travel distance calculator

```python
speed = float(input("Enter speed in km/h: "))
time = float(input("Enter travel time in hours: "))

distance = speed * time

print("Distance Travelled =", distance, "km")
```

8. BMI calculator

```python
height = float(input("Enter height in metres: "))
weight = float(input("Enter weight in kilograms: "))

bmi = weight / (height * height)

print("BMI =", bmi)
```

9. electricity bill estimator

```python
units = float(input("Enter units consumed: "))
cost_per_unit = float(input("Enter cost per unit: "))

bill = units * cost_per_unit

print("Estimated Electricity Bill = ₹", bill)
```

10. shopping receipt

```python
product1 = input("Enter first product name: ")
price1 = float(input("Enter price: "))

product2 = input("Enter second product name: ")
price2 = float(input("Enter price: "))

product3 = input("Enter third product name: ")
price3 = float(input("Enter price: "))

total = price1 + price2 + price3

print("\n--- Shopping Receipt ---")
print(product1, "₹", price1)
print(product2, "₹", price2)
print(product3, "₹", price3)
print("Total Purchase Amount = ₹", total)
```
