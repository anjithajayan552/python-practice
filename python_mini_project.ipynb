```python
item = input("Item: ")
price = float(input("Price: "))
quantity = int(input("Quantity: "))
member = input("Are you a member? (yes/no): ")

if price <= 0 or quantity <= 0:
    print("Error: Price and quantity must be greater than 0.")

else:
    # Subtotal
    subtotal = price * quantity

    # Discount
    if subtotal >= 5000:
        discount_rate = 10
    elif subtotal >= 2000:
        discount_rate = 5
    else:
        discount_rate = 0

    discount = subtotal * discount_rate / 100

    # Extra 2% for members
    if member == "yes":
        extra_discount = subtotal * 2 / 100
    else:
        extra_discount = 0

    total_discount = discount + extra_discount
    amount_after_discount = subtotal - total_discount

    # GST
    gst = amount_after_discount * 18 / 100

    # Total before delivery
    total = amount_after_discount + gst

    # Delivery charge
    if total > 1000:
        delivery = 0
        delivery_charge = "Free"
    else:
        delivery = 50
        delivery_charge = "50.00"

    # Final total
    total = total + delivery

    # final bill
    print(f"Item ={item}")
    print(f"Subtotal ={subtotal:.2f}")
    print(f"Discount  ={total_discount:.2f}")
    print(f"GST (18%)  = {gst:.2f}")
    print(f"Delivery   = {delivery_charge}")
    print(f"TOTAL   = {total:.2f}")
```
