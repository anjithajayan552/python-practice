28. password strength checker

```python

password = input("enter password:")

length=len(password)
uppercase=any(c.isupper() for c in password)
lowercase=any(c.isupper() for c in password)
digit=any(c.isdigit() for c in password)

if length <6:
  print("weak")
elif length <=9:
  print("medium")
elif uppercase and lowercase and digit:
  print("strong")
else:
  print("medium")
```
